```python
# Introduce the user to the calculator
print("Welcome user to the Modular Calculator Mini-Program!")

# Defines the operations so the program knows what operations to perform when asked by the user
def add_numbers(num1,num2):
  return (num1 + num2)
def subtract_numbers(num1,num2):
  return(num1 - num2)
def multiply_numbers(num1,num2):
  return (num1 * num2)
def divide_numbers(num1,num2):
  return (num1 / num2)

# Ask the user to enter the first number
first_number = int(input("Enter the first number:"))

# Ask the user to enter the second number
second_number = int(input("Second number:"))

# Ask the user to choose which operation to perform
print("Choose your operation:")
print("1 - Addition")
print("2 - Subtraction")
print("3 - Multiplication")
print("4 - Division")

# Letting the program know if the user chose an operation correctly
opr = int(input("Operation from 1-4: "))
if 1 <= opr <= 4:
 print("You successfully chose an opertion!")
else:
 print("Please put an operation that are provided!")

 # Let's the program know what operation is being used
 if opr == 1:
  print("You have chosen addition!")
 elif opr == 2:
  print("You have chosen subtraction!")
 elif opr == 3:
  print("You have chosen multiplication!")
 elif opr == 4:
  print("You have chosen division!")

# Solves the equation using the operations chosen by the user
if opr == 1:
  print(f"{first_number} + {second_number}")
  result = add_numbers(first_number, second_number)
  print(f"Result: {result}")
elif opr == 2:
  print(f"{first_number} - {second_number} ")
  result = subtract_numbers(first_number, second_number)
  print(f"Result: {result}")
elif opr == 3:
  print(f"{first_number} * {second_number}")
  result = multiply_numbers(first_number, second_number)
  print(f"Result: {result}")
elif opr == 4:
  print(f"{first_number} / {second_number}")
  result = divide_numbers(first_number, second_number)
  print(f"Result: {result} ")
```
